```python
import random

def rock_paper_scissors():
  choices = ["rock", "paper", "scissor"]
  computer_score = 0
  player_score = 0

  print("\n===Welcome to ROCK, PAPER, SCISSORS (First to 3)!===")

  while player_score < 3 and computer_score < 3:
    print(f"\n--- Current Score ---")
    print(f"Your score: {player_score} | Computer score: {computer_score}")

    computer = random.choice(choices)
    player = None

    while player not in choices:
       player = input("rock, paper or scissor? \n").strip().lower()
       if player not in choices:
        print("Invalid input. Please try again.")

    print(f"Computer chose: {computer}")
    if player == computer:
      print("It's a Tie!\n")
    elif player == "rock":
      if computer == "paper":
        print("You Lose!\n")
        computer_score += 1
      elif computer == "scissor":
        print("You Win!\n")
        player_score += 1
    elif player == "paper":
      if computer == "rock":
        print("You Win!\n")
        player_score += 1
      elif computer == "scissor":
        print("You Lose!\n")
        computer_score += 1
    elif player == "scissor":
      if computer == "paper":
        print("You Win!\n")
        player_score += 1
      elif computer == "rock":
        print("You Lose!")
        computer_score += 1

  if player_score == 3:
    print(f"\nCongratulations! You won the match with a score of {player_score} to {computer_score}!")
  else:
    print(f"\nYou lost! The computer won the match with a score of {computer_score} to {player_score}.")

while True:
  rock_paper_scissors()

  play_again = input("Do you want to play again? (yes/no): ").strip().lower()
  if play_again == "no":
     print("Thank you for playing!")
     break
```
