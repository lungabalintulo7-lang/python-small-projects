```python
import random

words = ("apple", "guitar", "ocean", "castle", "dragon",
    "camera", "rocket", "forest", "bridge", "planet",
    "silver", "winter", "tunnel", "mirror", "island",
    "shadow", "diamond", "thunder", "garden", "volcano",
    "desert", "pirate", "helmet", "jacket", "jungle",
    "laptop", "window", "hammer", "trophy", "elephant")

hang_man = {
    0: ("   ", "   ", "   "),
    1: (" O ", "   ", "   "),
    2: (" O ", " | ", "   "),
    3: (" O  ", "/|  ", "    "),
    4: (" O  ", "/|\\", "    "),
    5: (" O  ", "/|\\", "/   "),
    6: (" O  ", "/|\\", "/ \\")
}

def display_name(wrong_guesses):
    print("====================")
    for line in hang_man[wrong_guesses]:
        print(line)
    print("====================")

def display_hint(hint):
    print(" ".join(hint))

def display_answer(answer):
    print(" ".join(answer))

def main():
    playing = True
    while playing:
        answer = random.choice(words)
        hint = ["_"] * len(answer)
        wrong_guesses = 0
        guessed_letters = set()
        is_running = True

        while is_running:
            display_name(wrong_guesses)
            display_hint(hint)
            guess = input("Guess the letter: " ).lower()

            if len(guess) != 1 or not guess.isalpha():
                print("Enter a valid letter.")
                continue
            if guess in guessed_letters:
                print(f"{guess} has already been guessed.")
                continue

            guessed_letters.add(guess)
            if guess in answer:
                for index, letter in enumerate(answer):
                    if letter == guess:
                        hint[index] = guess
            else:
                wrong_guesses += 1

            if "_" not in hint:
                display_name(wrong_guesses)
                display_answer(answer)
                print("CONGRATULATIONS YOU WON!")
                is_running = False
            elif wrong_guesses == 6:
                display_name(wrong_guesses)
                display_answer(answer)
                print(f"You lost. Your word was {answer}")
                is_running = False

        play_again = input("Do you want to play again? (y/n) " ).lower()
        if play_again not in ("y", "yes"):
            playing = False
            print("Thank you for playing!")

if __name__ == "__main__":
    main()
```
