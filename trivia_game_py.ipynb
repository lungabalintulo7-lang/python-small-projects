```python
import random

questions = {
    "What color do you get when you mix blue and yellow?": "green",
    "What planet is closest to the Sun?": "mercury",
    "What is the chemical symbol for gold?": "Au",
    "In which country is the Great Pyramid of Giza located?": "egypt",
    "What is the capital city of France?": "paris",
    "What is the largest mammal on Earth?": "whale",
    "What gas do plants absorb from the atmosphere for photosynthesis?": "carbon",
    "What is the hardest natural substance on Earth?": "diamond",
    "What is the main ingredient in traditional guacamole?": "avocado",
    "Which ocean is the largest on Earth?": "pacific"
}

questions_list = list(questions.keys())
print(questions_list)

total_questions = 10
score = 0
selected_questions = random.sample(questions_list, total_questions)

for idx, question in enumerate(selected_questions):
    print(f"Question {idx + 1}: {question}")
    user_answer = input("Your answer: ").lower().strip()
    correct_answer = questions[question]
    if user_answer == questions[question].lower():
        print("Correct!\n")
        score += 1
    else:
        print(f"Incorrect. The correct answer is: {correct_answer}.\n")

print(f"Your final score is {score}/{total_questions}")

if score >= 4:
    print(f"You passed! Congratulations.")
else:
    print(f"You failed. Retry the game.")
```
