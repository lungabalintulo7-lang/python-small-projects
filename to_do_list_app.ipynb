```python
tasks = []

def show_menu():
  print("\n === To Do List ===")
  print("1. Add task")
  print("2. View tasks")
  print("3. Mark task as done")
  print("4. Delete task")
  print("5. Exit")

def add_task():
  task = input("Enter a task:\n")
  tasks.append({"task": task, "done": False})
  print(f"{task} has been added.")

def view_task():
  if not tasks:
    print("No tasks in the list.\n")
    return
  for index, t in enumerate(tasks, start=1):
     status = "✅" if t["done"] else "❌"
     print(f"{index}. {t['task']} [{status}]")

def mark_task_as_done():
  view_task()
  if not tasks:
    return
  try:
    index = int(input("Enter the task number to mark as done:\n")) - 1
    if 0 <= index < len(tasks):
      tasks[index]["done"] = True
      print("Task has been marked as done.\n")
    else:
      print("Invalid number.\n")
  except ValueError:
    print("Invalid Input. Please enter a valid number.")

def delete_task():
  view_task()
  if not tasks:
    return
  try:
    index = int(input("Enter the task number to delete:\n")) - 1
    if 0 <= index < len(tasks):
      del tasks[index]
      print("Task has been deleted.\n")
    else:
      print("Invalid number.\n")
  except ValueError:
    print("Invalid input. Please enter a number.\n")

while True:
  show_menu()
  choice = input("Choose option from 1 - 5:\n")
  if choice == "1":
    add_task()
  elif choice == "2":
    view_task()
  elif choice == "3":
    mark_task_as_done()
  elif choice == "4":
    delete_task()
  elif choice == "5":
    print("Goodbye!\n")
    break
  else:
    print("Invalid choice. Please choose a number from 1 - 5.\n")
```
