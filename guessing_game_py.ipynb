```python
import random

def number_guessing():
  print("Welcome to Lunga's Guessing Game!")
  print("I am thinking of a number from 1 to 40.\n")

  secret_number = random.randint(1, 40)
  attempts = 0
  max_attempts = 8

  while attempts < max_attempts:
    remaining_attempts = max_attempts - attempts
    print(f"Attempts remaining: {remaining_attempts}")

    try:
      guess = int(input("Take a guess from 1 - 40: ").strip())
      attempts += 1

      if guess > secret_number:
        print("Too high. Try again.\n")
      elif guess < secret_number:
        print("Too low. Try again.\n")
      else:
        print(f"Congratulations🎉! You guessed it in {attempts} attempts!\n")
        return True
    except ValueError:
      print("Please enter a valid number.\n")

  print(f"Game Over! You ran out of attempts. The secret number was {secret_number}.\n")
  return False

while True:
  number_guessing()
  choice = input("Would you like to play again? (yes/no) ").strip().lower()
  if choice == "no":
    print("Thank you for playing the game!")
    break
```
