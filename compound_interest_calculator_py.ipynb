```python
principle = 0
time = 0
rate = 0

while principle <= 0:
  try:
    principle = float(input("Enter the original amount of money invested, deposited, or borrowed: "))
    if principle <= 0:
      print("Principle can't be less than or equal to zero.")
  except ValueError:
    print("Invalid input. Please enter a valid number.")

while time <= 0:
  try:
    time = int(input("Enter the years: "))
    if time <= 0:
      print("Time can not be less than or equal to zero.")
  except ValueError:
    print("Invalid input. Please enter a valid integer.")

while rate <= 0:
  try:
    rate = float(input("Enter the interest rate: "))
    if rate <= 0:
      print("Interest rate should not be less than or equal to zero.")
  except ValueError:
    print("Invalid input. Please enter a valid number.")

total = principle * pow((1 + rate / 100), time)
print(f"Your balance after {time} year/s: R{total:.2f}")
```
