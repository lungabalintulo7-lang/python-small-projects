```python
import random
import string

def generate_password():
    length = int(input("How long would you like your password to be? (Your answer should be numeric value e.g 20.) ").strip())
    include_special = input("Would you like special characters in your password? (yes/no) ").strip().lower()
    include_uppercase = input("Would you like uppercase in your password? (yes/no) ").strip().lower()
    include_digits = input("Would you like numbers in your password? (yes/no) ").strip().lower()


    if length < 4:
      print(f"Password length must be atleast 4. ")
      return

    lowercase = string.ascii_lowercase
    uppercase = string.ascii_uppercase if include_uppercase == "yes" else ""
    digits = string.digits if include_digits == "yes" else ""
    special = string.punctuation if include_special == "yes" else ""

    all_characters = lowercase + special + digits + uppercase

    password = "".join(random.choice(all_characters) for _ in range(length))
    print(f"Generated password: {password}")
while True:
  generate_password()
  choice = input("Would you like to generate another password? (yes/no) ").strip().lower()
  if choice == "no":
    print("Exiting password generator. Goodbye!")
    break
```
